```python
from scipy.io import loadmat
import numpy as np
import matplotlib.pyplot as plt

# Path to one MAT file
FILE_PATH = r"C:\Users\Asus\OneDrive\Desktop\Projects\CWRU\CWRU\raw\B007_1_123.mat"

# Load MAT file
mat = loadmat(FILE_PATH)

# Display available variables
print("\nVariables inside the MAT file:")
for key in mat.keys():
    if not key.startswith("__"):
        print(f" - {key}")

# Automatically locate the Drive End signal
de_key = [key for key in mat.keys() if key.endswith("_DE_time")][0]

print(f"\nDrive End Signal Key: {de_key}")

# Extract signal
signal = mat[de_key].flatten()

print(f"Signal Shape : {signal.shape}")
print(f"Data Type    : {signal.dtype}")
print(f"Total Samples: {len(signal)}")

print("\nFirst 20 Samples:")
print(signal[:20])

# Basic Statistics
print("\nSignal Statistics")
print("----------------------------")
print(f"Minimum : {signal.min():.5f}")
print(f"Maximum : {signal.max():.5f}")
print(f"Mean    : {signal.mean():.5f}")
print(f"Std Dev : {signal.std():.5f}")

# Plot first 5000 samples
plt.figure(figsize=(15,5))
plt.plot(signal[:])
plt.title("B007 Bearing Fault - First 5000 Samples")
plt.xlabel("Sample Number")
plt.ylabel("Amplitude")
plt.grid(True)

plt.show()
```
